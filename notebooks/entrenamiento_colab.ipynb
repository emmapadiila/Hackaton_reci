# EcoConecta · entrenamiento reproducible
Selecciona GPU en **Entorno de ejecución**. El notebook usa los mismos scripts que el proyecto local. Los resultados se generan al ejecutar las celdas.

```python
from pathlib import Path
import os, subprocess, sys

# Desde Colab, introduce la URL Git que contiene esta versión de los scripts.
if not Path('train.py').is_file():
    repo = input('URL Git del proyecto EcoConecta: ').strip()
    subprocess.run(['git', 'clone', repo, '/content/ecoconecta'], check=True)
    os.chdir('/content/ecoconecta')
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'], check=True)
```

```python
data = Path('data/kaggle')
if not any(data.rglob('TRAIN')):
    data.mkdir(parents=True, exist_ok=True)
    archive = 'data/waste-classification-data.zip'
    if not Path(archive).is_file():
        subprocess.run(['curl', '-fL', 'https://www.kaggle.com/api/v1/datasets/download/techsash/waste-classification-data', '-o', archive], check=True)
    subprocess.run([sys.executable, 'waste_data.py', str(archive), str(data)], check=True)
```

```python
subprocess.run([sys.executable, 'train.py', '--epochs', '12'], check=True)
```

```python
from IPython.display import display, Image
import json
result = json.loads(Path('models/comparison.json').read_text())
for architecture in ('cnn', 'mobilenetv2'):
    display(Image(filename=f'models/{architecture}_curvas.png'))
    display(Image(filename=f'models/{architecture}_prueba.png'))
print(json.dumps(result, indent=2))
```
